4. Important Array Properties

5. Indexing

```python
marks = np.array([50,70,90,20,60])
print(marks[0])
print(marks[-4])
```

6. Indexing in a 2D Array

```python
marks = np.array([
    [50,70,40,60,20],
    [30,20,70,80,50]
    ])
print(marks[0, 2])
print(marks[1, -2])
```

7. Slicing

```python
marks = np.array([40,60,20,30,50])
print(marks[1:4])

marks = np.array([
    [40,60,20,30,50],
    [30,20,80,70,90]
    ])
print(marks[1,1:4])
```

8. Vectorized Operations

```python
marks = np.array([70,40,50,90,80])
print(marks + 5)
print(marks - 5)
print(marks * 5)
print(marks / 5)
```

9. Operations Between Two Arrays

```python
std1_marks = np.array([60,30,50,70,80])
std2_marks = np.array([40,20,70,30,90])

print(std1_marks + std2_marks)
print(std1_marks - std2_marks)
print(std1_marks * std2_marks)
print(std1_marks / std2_marks)
```

10. Mathematical Functions

```python
std1_marks = np.array([70,60,30,80,10])
print(np.sqrt(std1_marks))
print(np.power(std1_marks , 2))
std2_marks = np.array([40,-90,20,-70,30])
print(np.abs(std2_marks))
std3_marks = np.array([40.9,50.6,39.0,70.5,90.8])
print(np.round(std3_marks))
```

11. Aggregation Functions

```python
import numpy as np

marks = np.array([70, 85, 60, 95, 80])

print("Marks:", marks)

print("Sum:", np.sum(marks))
print("Average:", np.mean(marks))
print("Maximum Value:", np.max(marks))
print("Minimum Value:", np.min(marks))
print("Median:", np.median(marks))
print("Standard Deviation:", np.std(marks))
```

12. Conditions in NumPy

```python
import numpy as np

marks = np.array([30, 50, 80, 45, 90, 20])

print("Marks:", marks)

print("Marks >= 50:")
print(marks >= 50)

print("Marks > 50:")
print(marks > 50)

print("Marks < 50:")
print(marks < 50)

print("Marks == 50:")
print(marks == 50)
```

13. Filtering Data

```python
import numpy as np

marks = np.array([20, 40, 50, 45, 80, 70])

passed = marks[marks >= 50]
print(passed)

failed = marks[marks < 50]
print(failed)
```
